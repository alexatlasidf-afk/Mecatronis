```python

import os

with open('Second_file_test.txt','w') as p:
    p.write('Lalalalala')
    p.write('it goes aroung the world')
with open('Second_file_test.txt','a') as p:
    p.write('llalalalalala')

#write does an overide over other write
with open('Second_file_test.txt', 'w') as l:
    for i in range(101):
        l.write(f"{i}, {i**2}\n")

#פעולת OS מבצעת אוטומטית את הקובץ
os.startfile('Second_file_test.txt') 

```

```python

import os
#csv - stands for "comma seperated values"
#csv - opens with excel - it recognizes , as another column
with open('Second_file_test.csv', 'w') as l:
    l.write('x,y\n')
    for i in range(101):
        l.write(f"{i}, {i**2}\n")

#פעולת OS מבצעת אוטומטית את הקובץ
os.startfile('Second_file_test.csv') 
```

```python
x_values = []
y_values = []

#פעולת ה"נקסט" קופצץ על הושרה הראשונה - במקרה זה על הכותרת
#אם נרצה לקפוץ כמה שורות, נרשום כמה פעמים "נקסט" שנרצה..

with open('Second_file_test.csv','r') as f:
    next(f)
    for line in f:
        parts = line.split(',')
        x_values.append(int(parts[0]))
        y_values.append(int(parts[1]))
        
```

```python

import matplotlib.pyplot as plt

fig, ax = plt.subplots()
#plot sketches the line
ax.plot(x_values, y_values, linewidth = 5, color = "blue",linestyle = "--",)
ax.set_facecolor("lightblue")
#ereases the top and right frame lines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.show()

```
