```python
#"class" for Tal means like a bluprint when doing a project
#"קלאס" - זה המחלקה
#"state" - אסופה של כל הערכים של המשתנים
#OOP - Object Oriented Programming
#"class" - מתאר את הדבר בכללית
#דוגמה: נניח "קלרס" של חתול - בונה איך כל חתול צריך להיות
#כח ה"חתולים" בעולם שאנו הולכים לייצר יתנהגו ככה
```

```python
class Cat:
    
    #ככה נראית הפונקציה קבועה
    #double score fun "_ _" - "dunder"

    #"self" - פרמטר מיוחד שמגדיר את עצמו לפי ה"חתול" שנגדיר
    #נניח נבנה גם  חתול_2 אז "סלפ" יהפוך ל"חתול_2"

    #def __init__(self):
    #    self.name = "Miki"
    #    self.color = "black"

#"state" - Matsav, here "name" and "color" are states: 
#we will work only with "self" this course

#כל הפונקציות הנ"ל נקראות מטודות, כל מה שמתחת ל-קלאס

    def __init__(self, name, color):
        self.name = name
        self.color = color
        self.pat_counter = 0
    
    #פונקציה שמלטפת
    def pat(self):
        self.pat_counter += 1
        if self.pat_counter % 3 == 0:
            print("prrr prrr")
        else:
            print("...")

    def speak(self):
        print(f'{self.name} meows.')
        
```

```python
cat1 = Cat("Celine", "pink") #זה קריאת פונקציה של init
cat2 = Cat("Luna", "black") # זה האובייקטים
```

```python
cat1.name #Shows us cat1 name, based on Class
cat2.name
cat1.color
cat1.pat_counter #לא לשכוח לקמפל לפני שימוש

cat1.pat()
print(cat1.pat_counter)

cat2.speak()
```
