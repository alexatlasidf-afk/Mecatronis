```python
class Animal:
    
    def __init__(self, name, color):
        self.name = name
        self.color = color
        self.pat_counter = 0


    def pat(self):
        self.pat_counter += 1

    def speak(self):
        print(" * ")
```

```python
#Cat inherited the innet from animal
#נרצה לעשות פקוד ספיק שונה לחתול ספציפי משאר החיות

class Cat(Animal):

#We did an OVERRIDE for speak function
#יעבוד הספיק של החתול ספציפית ולא של חיות
    def speak(self):
        print(f"{self.name} meows")

    def pat(self):

        #"super class" - Animal
        # "sub class" - Cat
        #אנחנו רוצים להשתמש בפונ המקורית, ובעצם רק הוספנו פעולה
        
        super().pat(self) 
        if self.pat_counter % 3 == 0:
            print("prrrr")
        else:
            print("...")
    

```

```python
cat1 = Cat("Moshe", "Brown")
cat1.pat()
cat1.speak()
print(cat1.pat_counter)

```

```python
class Dog(Animal):

    #הוספנו את מספר צ'יפ של הכלב
    #בשימוש בסופר הגדרנו כי אנו מעניינית בסופר של חיות בצבע ושם

    def __init__(self, name, color, chip_id):
        super().__init__(name,color)
        self.dog_id = chip_id

    def pat(self):
        super().pat()
        if self.pat_counter % 2 == 0:
            print(" Happy Dog ")  
        else:
            print(f'{self.name} begs for more ')      


    def speak(self):
        print("Hav Hav")

```

```python
dog1 = Dog("Dylan","Beige","Unknown")
```

```python


dog1.pat()
print(dog1.pat_counter)
dog1.speak()
```
